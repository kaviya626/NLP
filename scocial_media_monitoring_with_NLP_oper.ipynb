```python
import pandas as pd
import nltk
from nltk.corpus import stopwords
import re
from collections import Counter

nltk.download('stopwords')
nltk.download('punkt')

df = pd.read_csv('/content/Text_Emotion.csv.zip')

stop_words = set(stopwords.words('english'))

def clean_text(text):
    text = text.lower()
    text = re.sub(r'http\S+|www\S+|https\S+', '', text)
    text = re.sub(r'\@w+|\#', '', text)
    text = re.sub(r'[^A-Za-z\s]', '', text)
    words = [word for word in text.split() if word not in stop_words]
    return words

df['clean_text'] = df['text'].apply(clean_text)

emotion_dict = {
    'happy': ['happy', 'joy', 'glad', 'pleased', 'delighted', 'smile', 'good', 'great', 'love'],
    'sad': ['sad', 'cry', 'unhappy', 'depressed', 'down', 'miserable'],
    'angry': ['angry', 'mad', 'furious', 'annoyed', 'hate', 'irritated'],
    'fear': ['fear', 'scared', 'afraid', 'terrified', 'nervous', 'worried'],
    'surprise': ['surprised', 'shocked', 'amazed', 'astonished'],
    'disgust': ['disgusted', 'gross', 'repulsed', 'sick', 'nauseous']
}

def detect_emotion(words):
    emotion_score = {emotion: 0 for emotion in emotion_dict}
    for word in words:
        for emotion, keywords in emotion_dict.items():
            if word in keywords:
                emotion_score[emotion] += 1
    if all(score == 0 for score in emotion_score.values()):
        return "neutral"
    return max(emotion_score, key=emotion_score.get)

df['predicted_emotion'] = df['clean_text'].apply(detect_emotion)


print(df[['text', 'predicted_emotion']].head(10))

def monitor_emotion(text):
    words = clean_text(text)
    predicted = detect_emotion(words)
    print(f"Text: {text}")
    print(f"Predicted Emotion: {predicted}")

monitor_emotion("I can't believe how amazing this day has been!")
monitor_emotion("This is so frustrating and annoying.")
```
